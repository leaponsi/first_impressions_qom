# Main — Python pipeline entry point

This notebook is the single entry point for the Python part of the project.
It installs dependencies, then runs the full pipeline in order:

1. `notebooks/01_exploration.ipynb` — visual verification on one participant
2. `notebooks/02_batch.ipynb` — batch pipeline for all 40 participants → `results/metrics.csv`

**For statistical analyses, see `main.Rmd`.**

## 0. Dependencies

```python
# Install required packages if not already available
import subprocess, sys

packages = ["numpy", "pandas", "scipy", "matplotlib", "nbconvert", "ipykernel"]

for pkg in packages:
    subprocess.run([sys.executable, "-m", "pip", "install", pkg, "-q"], check=True)

print("All dependencies installed.")
```

## 1. Run notebooks in order

```python
import os
import subprocess
import sys

ROOT      = os.path.abspath(os.getcwd())
notebooks = [
    os.path.join(ROOT, "notebooks", "01_exploration.ipynb"),
    os.path.join(ROOT, "notebooks", "02_batch.ipynb"),
]

for nb_path in notebooks:
    nb_name = os.path.basename(nb_path)
    print(f"Running {nb_name} ...")
    result = subprocess.run(
        [
            sys.executable, "-m", "nbconvert",
            "--to", "notebook",
            "--execute",
            "--inplace",
            "--ExecutePreprocessor.timeout=300",
            nb_path,
        ],
        capture_output=True,
        text=True,
    )
    if result.returncode == 0:
        print(f"  [OK] {nb_name}")
    else:
        print(f"  [ERROR] {nb_name}")
        print(result.stderr[-500:])  # show last 500 chars of error

print("\nDone.")
```

## 2. Verify output

```python
import pandas as pd

metrics_path = os.path.join(ROOT, "results", "metrics.csv")

if os.path.exists(metrics_path):
    df = pd.read_csv(metrics_path)
    print(f"metrics.csv produced successfully.")
    print(f"  Participants : {len(df)}")
    print(f"  Columns      : {len(df.columns)}")
    print(f"  Groups       :")
    print(df["group"].value_counts().to_string(index=True))
else:
    print("[ERROR] metrics.csv not found — check 02_batch.ipynb for errors.")
```
